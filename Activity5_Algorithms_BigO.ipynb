```python
LAST_NAME = "Mamac"
STUDENT_ID = "TUPM-26-4308"

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)

print("Seed:", SEED_DIGIT)
print("ID Sum:", ID_SUM)
print("Name Length:", NAME_LENGTH)
```

```python
data = [
    10 + SEED_DIGIT,
    25 - SEED_DIGIT,
    50 + SEED_DIGIT,
    5,
    100 - SEED_DIGIT
]

largest = data[0]
for num in data:
    if num > largest:
        largest = num

print("Dataset:", data)
print("Largest Value:", largest)
```
